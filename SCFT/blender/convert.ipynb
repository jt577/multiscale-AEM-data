```python
import torch
import numpy as np
from pathlib import Path

# Find all .pt files in current working directory
pt_files = sorted(Path(".").glob("*.pt"))

if not pt_files:
    print("No .pt files found in current directory.")

for pt_path in pt_files:
    name = pt_path.stem  # filename without .pt

    try:
        # Load tensor
        rho = torch.load(pt_path, map_location="cpu", weights_only=False)
        rho_numpy = rho.cpu().numpy()
        np.save(f"{name}.npy", rho_numpy)

    except Exception as e:
        print(f"❌ Failed to process {pt_path.name}: {e}")
```
