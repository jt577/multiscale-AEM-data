```python
import torch
import numpy as np
from pathlib import Path

# Find all .pt files in current working directory
pt_files = sorted(Path(".").glob("*.pt"))

if not pt_files:
    print("No .pt files found in current directory.")

for pt_path in pt_files:
    name = pt_path.stem  # filename without .pt

    try:
        # Load tensor
        rho = torch.load(pt_path, map_location="cpu", weights_only=False)
        rho_inv = 1 - rho

        # Ensure shape is (N, X, Y, Z)
        if rho.ndim != 4:
            rho = rho.unsqueeze(0)
            rho_inv = rho_inv.unsqueeze(0)

        rho_numpy = rho.cpu().numpy()
        rho_inv_numpy = rho_inv.cpu().numpy()

        # Stack original + inverse
        rho_full = np.zeros(
            (2, rho_numpy.shape[1], rho_numpy.shape[2], rho_numpy.shape[3]),
            dtype=rho_numpy.dtype,
        )
        rho_full[0] = rho_numpy
        rho_full[1] = rho_inv_numpy

        print(f"{pt_path.name} → {rho_full.shape}")

        np.save(f"{name}.npy", rho_full)

    except Exception as e:
        print(f"❌ Failed to process {pt_path.name}: {e}")
```
